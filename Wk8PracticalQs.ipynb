```python
#importing pandas
import pandas as pd
```

```python
#Q1: Exploring a new dataset
  #Read the csv file (OlympicsData.csv) and store to a DataFrame object

df = pd.read_csv("OlympicsData.csv")

#	Print out how many rows and columns are in the Data frame

print("The shape of the data set in terms of (rows, columns) is: ",df.shape)

#Print out the column headings contained in the Data Frame
print("The column headings include:")
df.columns

```

```python
#Q2: Create a copy of the dataframe object saved as a new dataframe object named SummerOlympics

SummerOlympics = df.copy()

#This DataFrame should only contain the data related to the summer Olympics. Therefore remove the columns
SummerOlympics.drop(['Winter Olympics', 'Gold.1', 'Silver.1', 'Bronze.1','Total.1', '# Games', 'Gold.2', 'Silver.2', 'Bronze.2', 'Combined Total'], axis=1, inplace=True)

#Display the SummerOlympics dataframe to check it contains the 6 columns and no others
SummerOlympics
```

```python
#Q3: Create another copy of the dataframe object saved as a new dataframe object named WinterOlympics

WinterOlympics = df.copy()

#This DataFrame should only contain the data related to the winter Olympics. Therefore remove the columns
WinterOlympics.drop([ 'Summer Olympics', 'Gold', 'Silver', 'Bronze', 'Total', '# Games', 'Gold.2', 'Silver.2', 'Bronze.2', 'Combined Total'], axis=1, inplace=True)

#Display the WinterOlympics dataframe to check it contains the 6 columns and no others
WinterOlympics
```

```python
#Q4: Create another copy of the original dataframe object saved as a new dataframe object named OlympicGames

OlympicGames = df.copy()

#This DataFrame should only contain the data associated with the columns highlighted:
OlympicGames.drop([ 'Summer Olympics', 'Gold', 'Silver', 'Bronze', 'Total', 'Winter Olympics', 'Gold.1', 'Silver.1', 'Bronze.1', 'Total.1'], axis=1, inplace=True)

#display
OlympicGames
```

```python
#Q5: 	You now have 3 different data frames, but they need further modifications. Each data frame should have the column labels as:
#Country, NoOfGames, NoOfGold, NoOfSilver, NoOfBronze, NoOfMedals


SummerOlympics.rename(columns={'Summer Olympics':'NoOfGames', 'Gold':'NoOfGold', 'Silver':'NoOfSilver', 'Bronze':'NoOfBronze', 'Total':'NoOfMedals'}, inplace=True)
WinterOlympics.rename(columns={'Winter Olympics':'NoOfGames', 'Gold.1':'NoOfGold', 'Silver.1':'NoOfSilver', 'Bronze.1':'NoOfBronze', 'Total.1':'NoOfMedals'}, inplace=True)
OlympicGames.rename(columns={'# Games':'NoOfGames', 'Gold.2':'NoOfGold', 'Silver.2':'NoOfSilver', 'Bronze.2':'NoOfBronze', 'Combined Total':'NoOfMedals'}, inplace=True)
```
