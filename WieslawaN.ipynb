```python
# Write a script that prompts the user to enter a Temperature in Fahrenheit.
# The entered value should be stored as a float.
# Convert the entered value into Celsius using the following formula:
# Celsius = (Fahrenheit - 32) * 5 / 9
# Print the temperature in Celsius.
# Example output:
# Please enter a Temperature in Fahrenheit 50
# The temperature 50.0 is 10.0

# Napisz skrypt, który prosi użytkownika o wprowadzenie temperatury w stopniach Fahrenheita.
# Wprowadzona wartość powinna zostać zapisana jako liczba zmiennoprzecinkowa (float).
# Przelicz wprowadzoną wartość na stopnie Celsjusza, używając następującego wzoru:
# Celsjusz = (Fahrenheit - 32) * 5 / 9
# Wyświetl temperaturę w stopniach Celsjusza.
# Przykładowy wynik działania:
# Please enter a Temperature in Fahrenheit 50
# The temperature 50.0 is 10.0

fahrenheit = float(input("Enter a Temperature in Fahrenheit "))
celsius = (fahrenheit - 32) * 5 / 9
print("The temperature " + str(fahrenheit) + " is " + str(celsius))
```

```python

# a. Create a list of subjects to include maths, programming, databases and statistics
# b. Display the list of subjects
# c. Ask the user to enter their least favourite subject from this list
# d. Find the index of their least favourite subject
# e. Delete their least favourite subject from the list
# f. Print the list in reverse order

# a. Stwórz listę przedmiotów zawierającą: maths, programming, databases i statistics
# b. Wyświetl listę przedmiotów
# c. Poproś użytkownika o podanie jego najmniej lubianego przedmiotu z tej listy
# d. Znajdź indeks jego najmniej lubianego przedmiotu
# e. Usuń najmniej lubiany przedmiot z listy
# f. Wyświetl listę w odwróconej kolejności

subjects = ["maths", "programming", "databases", "statistics"]

print("Subjects available:", subjects)

least_fav = input("Enter your least favourite subject from the list: ").strip().lower()

subject_index = subjects.index(least_fav)

print(f"The index of '{least_fav}' is: {subject_index}")

del subjects[subject_index]

subjects.reverse()

print("The list in reverse order is:", subjects)
```

```python

# a. Create a list of ten animals (You may choose any animals)
# b. Ask the user to enter two numbers between 1 and 10 – save these as variables
# c. Print out the part of the list which contains animals in the positions given by the user (remember index values start at 0)
# d. Add another animal value to the list so that it is the 6th animal in the list
# e. Sort the list in alphabetical order and print it out
# f. Ask the user to enter an animal that is in the list and delete the animal the user enters
# g. Delete all animals from the list

# a. Stwórz listę zawierającą dziesięć zwierząt (możesz wybrać dowolne)
# b. Poproś użytkownika o podanie dwóch liczb od 1 do 10 – zapisz je jako zmienne
# c. Wyświetl tę część listy, która zawiera zwierzęta na pozycjach podanych przez użytkownika (pamiętaj, że indeksy zaczynają się od 0)
# d. Dodaj kolejne zwierzę do listy tak, aby stało się szóstym zwierzęciem na liście
# e. Posortuj listę alfabetycznie i ją wyświetl
# f. Poproś użytkownika o podanie zwierzęcia z listy i usuń je
# g. Usuń wszystkie zwierzęta z listy

animals = ["cat", "dog", "lion", "tiger", "bear", "monkey", "rabbit", "deer", "fox", "wolf"]
print("Initial list:", animals)

num1 = int(input("Enter the starting position (1-10): "))
num2 = int(input("Enter the ending position (1-10): "))

start = min(num1, num2) - 1
end = max(num1, num2)
print("Animals in selected positions:", animals[start:end])

animals.insert(5, "giraffe")
print("List after inserting giraffe at 6th position:", animals)

animals.sort()
print("Alphabetically sorted list:", animals)

animal_to_remove = input("Enter an animal from the list to delete: ").strip().lower()
if animal_to_remove in animals:
    animals.remove(animal_to_remove)
    print(f"List after deleting {animal_to_remove}:", animals)
else:
    print("That animal is not in the list.")

animals.clear()
print("Final cleared list:", animals)
```
