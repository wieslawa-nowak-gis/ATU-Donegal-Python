```python

# Store as a variable named sum the total of values 2, 4, 6, 8. Print out the variable sum.
# Zapisz jako zmienną 'sum' sumę wartości 2, 4, 6, 8. Wydrukuj zmienną sum.

sum = 2 + 4 + 6 + 8
print ( sum )
```

```python
# Store as a variable named avg the average of 22, 19, 18. Print out the variable avg.
# Zapisz jako zmienną 'avg' średnią z 22, 19, 18. Wydrukuj zmienną avg.

avg = (22 + 19 + 18) / 3
print(avg)
```

```python
# Store as a variable named div the division of 195 by 40. Print out the variable div.
# Zapisz jako zmienną 'div' wynik dzielenia 195 przez 40. Wydrukuj zmienną div.

div = 195 / 40
print(div)
```

```python
# Store as a variable named minutes a value of 171. Print out whole hours and remaining minutes.
# Zapisz jako 'minutes' wartość 171. Wydrukuj ile to pełnych godzin i ile minut zostało.

minutes = 171
hours = minutes // 60
remaining_minutes = minutes % 60
print ( hours )
print ( remaining_minutes )
```

```python
# Store as a variable named power the value of 2 to the power of 7 (2^7). Print out the variable power.
# Zapisz jako zmienną 'power' wartość 2 do potęgi 7. Wydrukuj zmienną power.W Pythonie potęgowanie to dwie gwiazdki **

power = 2 ** 7
print ( power )
```

```python
# Store as a variable named name the text 'ATU Donegal'. Print out the length of the string name.
# Zapisz jako 'name' tekst 'ATU Donegal'. Wydrukuj długość (liczbę znaków) tego tekstu-Wydrukuj długość tego tekstu (liczbę znaków) za pomocą funkcji len()

name = "ATU Donegal"
print(len(name))
```
