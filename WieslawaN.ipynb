```python

# a. Create a list of ten animals (You may choose any animals)
# b. Ask the user to enter two numbers between 1 and 10 – save these as variables
# c. Print out the part of the list which contains animals in the positions given by the user (remember index values start at 0)
# d. Add another animal value to the list so that it is the 6th animal in the list
# e. Sort the list in alphabetical order and print it out
# f. Ask the user to enter an animal that is in the list and delete the animal the user enters
# g. Delete all animals from the list

# a. Stwórz listę zawierającą dziesięć zwierząt (możesz wybrać dowolne)
# b. Poproś użytkownika o podanie dwóch liczb od 1 do 10 – zapisz je jako zmienne
# c. Wyświetl tę część listy, która zawiera zwierzęta na pozycjach podanych przez użytkownika (pamiętaj, że indeksy zaczynają się od 0)
# d. Dodaj kolejne zwierzę do listy tak, aby stało się szóstym zwierzęciem na liście
# e. Posortuj listę alfabetycznie i ją wyświetl
# f. Poproś użytkownika o podanie zwierzęcia z listy i usuń je
# g. Usuń wszystkie zwierzęta z listy

animals = ["cat", "dog", "lion", "tiger", "bear", "monkey", "rabbit", "deer", "fox", "wolf"]
print("Initial list:", animals)

num1 = int(input("Enter the starting position (1-10): "))
num2 = int(input("Enter the ending position (1-10): "))

start = min(num1, num2) - 1
end = max(num1, num2)
print("Animals in selected positions:", animals[start:end])

animals.insert(5, "giraffe")
print("List after inserting giraffe at 6th position:", animals)

animals.sort()
print("Alphabetically sorted list:", animals)

animal_to_remove = input("Enter an animal from the list to delete: ").strip().lower()
if animal_to_remove in animals:
    animals.remove(animal_to_remove)
    print(f"List after deleting {animal_to_remove}:", animals)
else:
    print("That animal is not in the list.")

animals.clear()
print("Final cleared list:", animals)
```

```python
# endswith(wartość) – Zwraca TAK, jeśli tekst kończy się na podaną wartość.
# startswith(wartość) – Zwraca TAK, jeśli tekst zaczyna się od podanej wartości.
# isalnum() – Zwraca TAK, jeśli tekst zawiera tylko litery i cyfry (bez spacji, wykrzykników itp.).
# isalpha() – Zwraca TAK, jeśli tekst zawiera tylko litery (A-Z, bez cyfr i spacji).
# isdecimal() – Zwraca TAK, jeśli tekst zawiera tylko zwykłe cyfry.
# isdigit() – Zwraca TAK, jeśli wszystkie znaki to cyfry (łapie też np. małe cyfry w indeksach górnych²).
# islower() – Zwraca TAK, jeśli wszystkie litery w tekście są małe (np. "kot").
# isupper() – Zwraca TAK, jeśli wszystkie litery w tekście są WIELKIE (np. "KOT").
# istitle() – Zwraca TAK, jeśli tekst zaczyna się od wielkiej litery, a reszta jest mała (jak Tytuł, np. "Python").


# 1. startswith(wartość) -> Szukamy linków z bezpiecznym początkiem "https://"
link = "https://youtube.com"
if link.startswith("https://"):
    print("TAK! Ta strona jest bezpieczna, bo zaczyna się od https://")

# 2. endswith(wartość) -> Sprawdzamy, czy plik to piosenka (ma końcówkę .mp3)
plik = "ulubiona_piosenka.mp3"
if plik.endswith(".mp3"):
    print("TAK! To jest plik muzyczny, bo kończy się na .mp3")

# 3. isalnum() -> Sprawdzamy nazwę gracza (dozwolone tylko litery i cyfry, bez spacji!)
nick_gracza = "Ninja2000"
if nick_gracza.isalnum():
    print("TAK! Nazwa gracza jest poprawna, zawiera tylko litery i cyfry.")

# 4. isalpha() -> Sprawdzamy kod rabatowy do sklepu (muszą być same litery)
kod_znizkowy = "SUPERZALIZKA"
if kod_znizkowy.isalpha():
    print("TAK! Kod jest poprawny, składa się z samych liter.")

# 5. isdecimal() -> Sprawdzamy wiek użytkownika (muszą być zwykłe cyfry)
wiek = "14"
if wiek.isdecimal():
    print("TAK! Wiek został wpisany poprawnie przy użyciu samych cyfr.")

# 6. isdigit() -> Sprawdzamy kod PIN do telefonu (same cyfry)
pin_telefonu = "4829"
if pin_telefonu.isdigit():
    print("TAK! Kod PIN składa się z samych cyfr.")

# 7. islower() -> Sprawdzamy adres e-mail (zazwyczaj pisze się go tylko małymi literami)
email = "moj.email@gmail.com"
if email.islower():
    print("TAK! Adres e-mail jest zapisany poprawnie samymi małymi literami.")

# 8. isupper() -> Wykrywacz krzyku internetowego (Caps Lock)
wiadomosc = "NIE CHCĘ JEŚĆ WARZYW"
if wiadomosc.isupper():
    print("TAK! Ta wiadomość jest napisana wielkimi literami (użytkownik krzyczy!).")

# 9. istitle() -> Sprawdzamy, czy nazwa miasta zaczyna się od Wielkiej Litery
miasto = "Warszawa"
if miasto.istitle():
    print("TAK! Nazwa miasta jest zapisana poprawnie (Wielka litera na starcie).")
```

```python
# A year is a leap year if it can be exactly divisible by 4 (no remainder)
# but not divisible by 100 (eg.. 2024 is a leap year). If a year can be
# divisible by 100 and 4, it must also be exactly divisible by 400 to be a
# leap year. (e.g 1600 is a leap year, 1900 is not a leap year)
# Develop a program asking for a year and determine if it is a leap year.

# Rok jest przestępny, jeśli dzieli się dokładnie przez 4 (bez reszty),
# ale nie dzieli się przez 100 (np. 2024 to rok przestępny). Jeśli rok dzieli się
# zarówno przez 100, jak i przez 4, musi również dzielić się dokładnie przez 400,
# aby był przestępny (np. 1600 to rok przestępny, ale 1900 nie jest przestępny).
# Stwórz program, który prosi o podanie roku i sprawdza, czy jest on przestępny.

year = int(input("Please enter a year: "))

# Logika roku przestępnego:
# Rok jest przestępny, jeśli dzieli się przez 400 ORAZ (dzieli się przez 4 i NIE dzieli się przez 100)
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

```python
word = "ATLANTIC"  # 1. Tajne słowo do odgadnięcia
dashes = ["_"] * len(word)  # 2. Tworzy listę podkreśleń (kresek) – tyle, ile jest liter w słowie
attempts = 6  # Liczba dozwolonych błędnych prób (odpowiada 6 częściom ciała wisielca)

while "_" in dashes and attempts > 0:  # PĘTLA: Powtarzaj, dopóki są ukryte litery I masz wolne próby
    print(" ".join(dashes))  # Wyświetla aktualny stan słowa z odstępami między kreskami
    guess = input("Guess a letter: ").upper()  # 3. Prosi o literę i automatycznie zmienia ją na wielką

    if guess in word:  # 4. INSTRUKCJA IF: Jeśli zgadnięta litera znajduje się w tajnym słowie
        for index, letter in enumerate(word):  # Przechodzi przez słowo, sprawdzając pozycję (index) i samą literę
            if letter == guess:  # Jeśli litera w słowie pasuje do tej podanej przez gracza
                dashes[index] = guess  # Zamienia kreskę na właściwą literę na tej samej pozycji
    else:  # W PRZECIWNYM RAZIE: Jeśli litery nie ma w słowie
        attempts -= 1  # Odejmuje 1 od liczby pozostałych prób (skrót od attempts = attempts - 1)
        print(f"Wrong! Remaining attempts: {attempts}")  # Informuje gracza, ile prób mu zostało

# KONIEC GRY: Uruchamia się automatycznie po wyjściu z pętli while
if "_" not in dashes:  # Jeśli w słowie nie ma już żadnych kresek (wszystko odgadnięte)
    print(f"Game over - the user has Won! The word was: {word}")  # Komunikat o wygranej
else:  # W przeciwnym razie (oznacza to, że liczba prób spadła do 0)
    print(f"Game over - the user has lost. The word was: {word}")  # Komunikat o przegranej
```

```python
def play_hangman(num):
    playWord = word_list(num-1)
    guesses = []
    lives = 6  # Number of attempts allowed (head, body, 2 arms, 2 legs)

    print("Welcome to Hangman!\n")

    while lives > 0:
        displayLetters = display_game(playWord, guesses)
        print(displayLetters)

        if check_word_found(displayLetters):
          lives = -1
        else:
          guess = input("\nPlease take a guess: ").lower()

          if guess in guesses:
            print("you have already guessed this letter")
          else:
            guesses.append(guess)

            if guess not in playWord:
              print("That letter is not there!\n")
              lives = lives - 1
              build_man(lives)
            else:
              print("The letter is there!")

          print("Your guesses: ", guesses)

    if lives == 0:
        print("GAME OVER - You guessed 6 wrong letters!. The word was:", playWord)
    elif lives == -1:
        print("GAME OVER - You guessed all the letters! YOU WON!")
```

```python
#example of a collection as parameter  (list)

def adding_numbers(numbers):
  sum = 0                   #variable initialised as 0 to hold running total
  for num in numbers:       #loop through list
    sum = sum + num         #add each number to the previous sum

  print("The sum of the list provided is ", sum)    # print result
```

```python
#Dictionary Looping incorporated into the function?

def adding_numbers(numbers):
  if(isinstance(numbers, dict)):    #isInstance is a python function to check if the collection is a dictionary (returns true or false)
    sum = 0
    for num in numbers.values():       #loop through dictionary VALUES
      sum = sum + num

  else:
    sum = 0                   #variable initialised as 0 to hold running total
    for num in numbers:       #loop through list
      sum = sum + num         #add each number to the previous sum

  print("The sum of the list provided is ", sum)    # print result
```

```python
numDICT = { 1: 14, 2:23, 3:46}
adding_numbers(numDICT) # adds the VALUES from the dictionary
```

```python
# Write a function which will convert a passed in temperature in Fahrenheit to Degrees Celsius and return the degrees value.
# Napisz funkcję, która przekonwertuje podaną temperaturę w stopniach Fahrenheita na stopnie Celsiusza i zwróci obliczoną wartość.

def fahrenheit_to_celsius(fahrenheit):
    celsius = (fahrenheit - 32) * 5 / 9
    return celsius
```

```python
# Write a function which will accept a parameter of a distance in miles, and return a converted kilometers value.
# Napisz funkcję, która przyjmie jako parametr odległość w milach i zwróci przeliczoną wartość w kilometrach.

def miles_to_kilometers(miles):
    kilometers = miles * 1.609344
    return kilometers
```

```python
# Write a script (not using a function) which asks the user whether they wish to convert temperature or distance.
# If the user chooses temperature – ask for a temperature to convert and use the function developed in Q1 to print out the associated conversion.
# If the user chooses distance – ask for a distance to convert and use the function developed in Q2 to print out the associated conversion.
# If the user enters a value that is not numeric, print out a more appropriate message and prevent the program from crashing.

# Napisz skrypt (bez użycia funkcji), który pyta użytkownika, czy chce przekonwertować temperaturę czy odległość.
# Jeśli użytkownik wybierze temperaturę – zapytaj o temperaturę do konwersji i użyj funkcji z pytania Q1, aby wyświetlić wynik.
# Jeśli użytkownik wybierze odległość – zapytaj o odległość do konwersji i użyj funkcji z pytania Q2, aby wyświetlić wynik.
# Jeśli użytkownik wprowadzi wartość, która nie jest liczbą, wyświetl odpowiedni komunikat i zapobiegnij awarii programu.

choice = input("What do you want to convert? Enter 'temperature' or 'distance': ").strip().lower()

if choice == 'temperature':
    user_input = input("Enter temperature in Fahrenheit: ")
    try:
        fahr = float(user_input)
        cels = fahrenheit_to_celsius(fahr)
        print(f"{fahr}°F is equal to {cels:.2f}°C")
    except ValueError:
        print("Error: Please enter a valid numeric value for temperature.")

elif choice == 'distance':
    user_input = input("Enter distance in miles: ")
    try:
        mi = float(user_input)
        km = miles_to_kilometers(mi)
        print(f"{mi} miles is equal to {km:.2f} kilometers")
    except ValueError:
        print("Error: Please enter a valid numeric value for distance.")

else:
    print("Invalid choice. Please restart the script and enter 'temperature' or 'distance'.")
```

```python
# Write a function called even_list – that accepts 2 parameters (a minimum number and a maximum number).
# The function should return a list of all the even numbers between the two parameters which are passed in.
# Write a script to test the function created. (Input does not have to be user given, but can be if you choose)

# Napisz funkcję o nazwie even_list, która przyjmuje 2 parametry (liczbę minimalną i maksymalną).
# Funkcja powinna zwrócić listę wszystkich liczb parzystych znajdujących się pomiędzy przekazanymi parametrami.
# Napisz skrypt testujący utworzoną funkcję. (Dane wejściowe nie muszą pochodzić od użytkownika, ale mogą, jeśli chcesz)

def even_list(min_num, max_num):
    evens = []
    for num in range(min_num, max_num + 1):
        if num % 2 == 0:
            evens.append(num)
    return evens

# Test script / Skrypt testujący
test_min = 1
test_max = 15
result = even_list(test_min, test_max)

print(f"Even numbers between {test_min} and {test_max}:")
print(result)
```

```python
# Write 2 functions which will manipulate lists.
# 1. The first function will create the list by asking the user to enter 5 different animals and return the list created.
# 2. The second function will display the list created from the first function and ask the user to add another animal. If the user enters an animal that is already in the list, an appropriate message should be given to the user and the animal is not added to the list and they are asked for another animal. If the user enters an animal not in the list, the animal should be added to the list and the list returned.
# 3. Write a script which calls the first function once and the second function 3 times then print out the list which would now have 8 animals.

# Napisz 2 funkcje, które będą manipulować listami.
# 1. Pierwsza funkcja utworzy listę, prosząc użytkownika o wpisanie 5 różnych zwierząt i zwróci utworzoną listę.
# 2. Druga funkcja wyświetli listę utworzoną przez pierwszą funkcję i poprosi użytkownika o dodanie kolejnego zwierzęcia. Jeśli użytkownik wpisze zwierzę, które jest już na liście, należy wyświetlić odpowiedni komunikat, zwierzę nie zostanie dodane, a użytkownik zostanie poproszony o podanie innego. Jeśli użytkownik wpisze zwierzę, którego nie ma na liście, powinno ono zostać dodane, a funkcja zwróci zaktualizowaną listę.
# 3. Napisz skrypt, który wywoła pierwszą funkcję raz, a drugą funkcję 3 razy, a następnie wypisze listę, która będzie teraz zawierać 8 zwierząt.

def create_animal_list():
    animals = []
    print("Please enter 5 different animals:")
    while len(animals) < 5:
        animal = input(f"Enter animal {len(animals) + 1}: ").strip().lower()
        if animal == "":
            print("Name cannot be empty. Try again.")
        elif animal in animals:
            print("You already entered that animal. Please enter a different one.")
        else:
            animals.append(animal)
    return animals

def add_animal(animal_list):
    print(f"\nCurrent list of animals: {animal_list}")
    while True:
        new_animal = input("Enter another animal to add to the list: ").strip().lower()
        if new_animal == "":
            print("Name cannot be empty. Try again.")
        elif new_animal in animal_list:
            print(f"The animal '{new_animal}' is already in the list. Please enter a different one.")
        else:
            animal_list.append(new_animal)
            print(f"'{new_animal}' has been successfully added!")
            return animal_list

# Test script / Skrypt testujący
# Call the first function once to get 5 animals
my_animals = create_animal_list()

# Call the second function 3 times to reach 8 animals
my_animals = add_animal(my_animals)
my_animals = add_animal(my_animals)
my_animals = add_animal(my_animals)

# Print out the final list of 8 animals
print("\nFinal list of 8 animals:")
print(my_animals)
```
