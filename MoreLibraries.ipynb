**Datetime Library**

*PPT Slide 3*

```python
#import the datetime library and use alias dt for easier reference
import datetime as dt
```

```python
#get todays date and save as datetime object todaysDate
todaysDate = dt.datetime.now()    #now() is a method to get the current date
print(todaysDate)
```

```python
#datetime properties
print("I can access the day (date 1->28/29/30/31) of the month using the DAY property: ", todaysDate.day)
print("I can access the month of the year (1-12) using the MONTH property: ", todaysDate.month)
print("I can access the year from the date using the YEAR property: ", todaysDate.year)
print("I can access the hour (24 hr format) from the time using the HOUR property: ", todaysDate.hour)
print("I can access the minute (up to 60) from the time using the MINUTE property: ", todaysDate.minute)
print("I can access the second (up to 60) from the time using the SECOND property: ", todaysDate.second)
print("I can access the microsecond (up to 1000000) from the time using the MICROSECOND property: ", todaysDate.microsecond)

#datetime methods
print("I can access the day of the month as an integer (0 = Monday, 6 = Sunday) using the WEEKDAY method: ", todaysDate.weekday())
print("I can access the date part only of a date time using the DATE method: ", todaysDate.date())
print("I can access the time part only of a date time using the TIME method: ", todaysDate.time())
print("I can also format my date as a string output to get the day of the week with %A argument giving day : ", todaysDate.strftime("%A"), "or %a argument giving day :", todaysDate.strftime("%a") )
print("I can also format my date as a string output to get the day of the week with %B argument giving month : ", todaysDate.strftime("%B"), "or %b argument giving month", todaysDate.strftime("%b"))
print("Todays is", todaysDate.strftime("%A, %d %B, %Y"))
print("Todays is", todaysDate.strftime("%d/%m/%y"))
```

```python
#accessing the date class we can import just the date from the datetime library
from datetime import date

anotherDate = date(2025, 12, 25)
print("This is another date object I created with the date: ", anotherDate.strftime("%d/%m/%y"))
```

```python
#accessing the time class we can import just the time from the datetime library
from datetime import time
myTime = time(23, 56, 22)
print("this is a time i created: ", myTime)
```

```python
#accessing the timedelta class we can use date and time math
from datetime import timedelta
today = dt.datetime.now()

tomorrow = today + timedelta(days=1)
in4hours = today + timedelta(hours = 4)
in1year = today + timedelta(days = 365)
yesterday = today - timedelta(days =1)

print(tomorrow)
print(in4hours)
print(in1year)
print(yesterday)
```

```python
#I can simply subtract to find difference in dates
diff = date(2024, 11, 18) - date(2024, 11, 15)
print(diff.days)  # Outputs 3
```

```python
#use timezone to set different timezones
from datetime import timezone

tz = timezone(timedelta(hours=10, minutes = 30))  # GMT+10:30
AdelaideAu = dt.datetime.now(tz)
print(AdelaideAu)
```

***array methods***

```python
#copy an array
array1 = np.array([1, 2, 3, 4, 5, 6])
array2 = array1.copy()
print(array2)
```

```python
#join arrays - concatenate
array3 = np.concatenate((array1, array2))
print(array3)
```

```python
#search - find index value of an array
vals = np.where(array3 == 4)
print(vals)
```

```python
#sort arrays
print(np.sort(array3))
```

```python
#perform aritmetic on arrays without loops
print(array1 + array2)
```

```python
#list would require a loop
list1 = [1, 2, 3, 4, 5, 6]
list2 = list1.copy()

for index in range(len(list1)):
  print(list1[index]+list2[index])
```

# ***Other Useful Modules***

**Random Module**
*PPT Slide 5*

```python
#import the random Module to use the radom class and methods
import random as r  #using alias r
```

```python
#default use of random method will generate a float between 0 and 1
num = r.random()
print(num)
```

```python
#randint method will return a value between two parameters (include starting value, include ending value)
intNum = r.randint(0, 100)
print(intNum)
```

```python
#randrange method will return a value between the range arguments but also can include steps...
rangedNum = r.randrange(0,100,10)
print(rangedNum)
```

```python
colourList = ["red", "green", "blue", "black", "yellow", "orange", "pink", "purple", "white"]

#Choice method can select a random item from a collection like a list
randomColour = r.choice(colourList)
print(randomColour)
```

Use of random could be useful to enhance the hangman game to generate a random number selected from the CSV file so the user would get a new word every time they played automatically

**Statistics Module**

```python
#import the statistics Module with an alias s
import statistics as s
```

```python
nums = np.array([1,2,3, 4, 5, 6, 6, 6, 7, 8, 8, 9, 10, 10, 12, 13, 15])
#example statistic methods
print("Mean: ", s.mean(nums.tolist()))
print("Mode: ", s.mode(nums.tolist()))
print("Median: ", s.median(nums.tolist()))
print("Standard Deviation: ", s.stdev(nums.tolist()))
print("Variance:", s.variance(nums.tolist()))
```

# **Pandas - includes many of these additional modules within their library**

```python
import pandas as pd

#no need for additional CSV library as PANDAS has a method
df = pd.read_csv("SalesData.csv")
df
```

**Pandas includes Math and Statistical Methods**

```python
#range of methods to perform math and statistics without using Math or statistics library
sales = df["Sales"]
print("Count: ", sales.count())
print("Max: ", sales.max())
print("Mean: ", sales.mean())
print("Median: ", sales.median())
print("Product: ", sales.prod())
print("Standard Deviation: ", sales.std())
print("Sum: ", sales.sum())
print("Value Counts: ", sales.value_counts())
print("Variance:", sales.var())
```

**Pandas includes visualisation - Plotting**

```python
#plot method to plot different charts from the dataframes and series data without using matplotlib

regionSales = df["Region"].value_counts()

regionSales.plot(kind='bar')
```

```python
df.groupby(['Region']).sum()
```

```python
dataToPlot = df.groupby(['Region']).sum()

dataToPlot.plot(kind="bar", title = "Sales / Expenses / Profits Per Region")
```
