I CAN CREATE A FUNCTION FOR ALL THIS "CLEANING"

```python
def dataPrep(data):
    countries = data["Country"]

    data[['Country', 'Code']] = data['Country'].str.extract(r'^(.*?)\s*[\(\[]([^\)\]]+)[\)\]]$')

    data['Country'] = data['Country'].str.strip().str.split("(").str[0]
    data['Code'] = data['Code'].str.strip()

    data['relative % win'] = round(data['NoOfMedals'] / data['NoOfGames'] * 100, 2)

    data.set_index(data["Code"], inplace=True)
    data.drop("Code", axis=1, inplace=True)

    data = data.reindex(columns=[
        'Country', 'Code', 'NoOfGames', 'NoOfGold',
        'NoOfSilver', 'NoOfBronze', 'NoOfMedals', 'relative % win'
    ])

    return data
```

```python
winterData = pd.read_csv("WinterOlympics.csv", index_col=0)
winterData
```

```python
NewWinterData = dataPrep(winterData)
NewWinterData
```

**Expression	Explanation**

r	Raw string. Tells Python not to interpret backslashes as escape characters.

^	Start of the string.

(.*?)	Capture group 1. Matches the country name.

\s*	Matches zero or more whitespace characters (spaces, tabs, etc.).

[\(\[]	Matches either ( or [.

([^\)\]]+)	Capture group 2. Matches one or more characters that are NOT ) or ].

[\)\]]	Matches either ) or ].

$	End of the string.
